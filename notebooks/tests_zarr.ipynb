```python
import rasterio

import rioxarray

import xarray as xr

import odc.geo.xr
from odc.geo import odc

import geoutils as gu

import zarr
```

## rasterio

```python
from rasterio.shutil import copy as rio_copy

with rasterio.open("../data/rasters/IMAGERY.tif") as src:
    profile = src.profile.copy()
    profile.update({
        "driver": "zarr",
        "blockxsize": 512,   # chunk size X
        "blockysize": 512,   # chunk size Y
        "chunks": (1, 512, 512),  # (bands, rows, cols)
    })

    rio_copy(src, "../outputs/rasterio.zarr", **profile)
```

## rioxarray

```python
da = rioxarray.open_rasterio("../data/rasters/IMAGERY.tif", chunks={"x": 512, "y": 512})
da.to_zarr("../outputs/rioxarray.zarr", mode="w")
```
